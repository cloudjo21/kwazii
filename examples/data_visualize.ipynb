```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Example data
df = pd.DataFrame({
    'Data1': [1, 2, 2, 3, 3, 3, 4, 4, 5],
    'Data2': [2, 3, 3, 4, 4, 4, 5, 6, 6],
    'Data3': [1, 3, 3, 3, 4, 5, 5, 6, 7]
})

# Reshape the DataFrame
df_melt = pd.melt(df, var_name='Dataset', value_name='Value')

# Creating a FacetGrid
g = sns.FacetGrid(df_melt, col='Dataset', col_wrap=3, sharex=False, sharey=False)

# Mapping the sns.histplot function to each subplot
g.map(sns.histplot, 'Value', kde=True, bins=10, alpha=0.5)

# Adding titles and labels
g.set_titles('{col_name}')
g.set_axis_labels('Value', 'Frequency')

# Adjust layout
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Example data
data1 = [1, 2, 2, 3, 3, 3, 4, 4, 5]
data2 = [2, 3, 3, 4, 4, 4, 5, 6, 6]
data3 = [1, 3, 3, 3, 4, 5, 5, 6, 7]

# List of datasets
data_list = [data1, data2, data3]

# Plotting
plt.figure(figsize=(10, 6))
for data in data_list:
    sns.histplot(data, kde=True, bins=10, alpha=0.5)  # alpha for transparency

plt.legend(['Data 1', 'Data 2', 'Data 3'])
plt.title('Multiple Histograms')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Example data
df = pd.DataFrame({
    'Data1': [1, 2, 2, 3, 3, 3, 4, 4, 5],
    'Data2': [2, 3, 3, 4, 4, 4, 5, 6, 6],
    'Data3': [1, 3, 3, 3, 4, 5, 5, 6, 7]
})

# Reshape the DataFrame
df_melt = pd.melt(df, var_name='Dataset', value_name='Value')

# Plotting
plt.figure(figsize=(10, 6))
sns.histplot(data=df_melt, x='Value', hue='Dataset', kde=True, bins=10, alpha=0.5)

plt.title('Multiple Histograms')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.show()
```
