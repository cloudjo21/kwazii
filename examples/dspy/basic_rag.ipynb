## Basic QA Prediction

```python
input_question = "{}\n{}".format(content, "연구팀의 논문은 어디서 발표되었나요?")
```

```python
qa = dspy.Predict('question: str -> response: str')
response = qa(question=input_question)

print(response.response)
```

```python
dspy.inspect_history(n=1)
```

## Basic CoT

```python
input_question = "{}\n\n{}".format(content, "연구팀이 미세플라스틱을 오염물질에 노출시키는 실험에서 어떤 현상을 가정하였나요? 그리고 앞으로의 연구는 어떤 실험을 준비하고 있나요?")
```

```python
cot = dspy.ChainOfThought('question -> response')
cot(question=input_question)
```

```python
dspy.inspect_history(n=2)
```

## Metrics for LM System Quality
- Evaluation Metrics
    - ground-truth labels of ideal outputs
        - e.g. for classification or question answering
    - self-supervised metrics
        - e.g. checking faithfulness or lack of halluciation
- Evaluation Metric for QA
  - How well does the system response cover all key facts in the gold response? And the other way around, how well is the system response not saying things that aren't in the gold response?
