```python
from agno.agent import Agent
from agno.tools.yfinance import YFinanceTools
from agno.models.ollama import Ollama, OllamaTools
from ollama import Client as OllamaClient

agent = Agent(
    # model=Ollama(
    #     id="tiger-gemma2",
    #     host="http://localhost:11434/api/chat"
    #     # client=OllamaClient(host="http://localhost:11434/api/chat"),
    # ),
    model=OllamaTools(
        id="tiger-gemma2",
        host="http://localhost:11434/api/chat",
        client=OllamaClient(host="http://localhost:11434/api/chat"),
    ),
    tools=[YFinanceTools(stock_price=True)],
    markdown=True,
)
agent.print_response("What is the stock price of Apple?", stream=True)
```

```python
from agno.tools.function import Function

def add(a: float, b: float) -> float:
    """
    두 숫자가 주어졌을 때 숫자의 합을 계산하고 반환한다.
    
    Args:
        a (float): 첫번째 숫자
        b (float): 두번째 숫자
    
    Returns:
        float: 두 숫자의 합
    """
    return a + b

add_tool = Function.from_callable(add)

agent = Agent(
    # model=Ollama(
    #     id="tiger-gemma2",
    #     host="http://localhost:11434/api/chat"
    #     # client=OllamaClient(host="http://localhost:11434/api/chat"),
    # ),
    model=Ollama(
        id="tiger-gemma2",
        host="http://localhost:11434/api/chat",
        # client=OllamaClient(host="http://localhost:11434/api/chat"),
    ),
    tools=[add_tool],
    markdown=True,
)
agent.print_response("What is the sum of 5 and 10?", stream=True)
```
